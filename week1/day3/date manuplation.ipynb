```python
%sql
SELECT 
  current_date() AS current_date,
  current_timestamp() AS current_timestamp,
  date_format(current_timestamp(), 'HH:mm:ss') AS current_time;
```

```python
%sql
SELECT 
  YEAR(order_date),
  MONTH(order_date),
  DAY(order_date)
FROM orders;
```

```python
%sql
SELECT 
  EXTRACT(YEAR FROM order_date),
  EXTRACT(MONTH FROM order_date),
  EXTRACT(DAY FROM order_date)
FROM orders;
```

```python
%sql
SELECT 
  MONTHNAME(order_date),
  DAYNAME(order_date)
FROM orders;
```

```python
%sql
SELECT 
  WEEKDAY(order_date),
  DAYOFWEEK(order_date)
FROM orders;
```

```python
%sql
-- Method 1
SELECT * FROM orders
WHERE DAYNAME(order_date) IN ('Saturday', 'Sunday');

-- Method 2
SELECT * FROM orders
WHERE DAYOFWEEK(order_date) IN (1, 7);
```

```python
%sql
SELECT * FROM orders
WHERE DAYOFWEEK(order_date) BETWEEN 2 AND 6;
```

```python
%sql
SELECT 
  order_date,
  DATE_ADD(order_date, 5) AS plus_5_days,
  DATE_SUB(order_date, 3) AS minus_3_days,
  ADD_MONTHS(order_date, 1) AS plus_1_month
FROM orders;
```

```python
%sql
SELECT 
  order_id,
  DATEDIFF(delivery_date, order_date) AS delivery_days
FROM orders;
```

```python
%sql
SELECT 
  TIMESTAMPDIFF(DAY, order_date, delivery_date) AS days_diff,
  TIMESTAMPDIFF(MONTH, order_date, delivery_date) AS months_diff
FROM orders;
```

```python
%sql
SELECT LAST_DAY(order_date) FROM orders;
```

```python
%sql
SELECT DATE_SUB(order_date, DAY(order_date) - 1) AS first_day
FROM orders;
```

```python
%sql
SELECT date_format(order_date, 'dd-MM-yyyy') FROM orders;

SELECT date_format(order_date, 'MMMM dd, yyyy') FROM orders;
```

```python
%sql
SELECT TO_DATE('15-01-2024', 'dd-MM-yyyy') AS converted_date;
```

```python
%sql
SELECT date_format(order_timestamp, 'dd-MM-yyyy HH:mm:ss')
FROM orders;
```

```python
%sql
SELECT * FROM orders 
WHERE date_format(order_date, 'MMMM') = 'February';
```

```python
%sql
SELECT 
  order_date,
  CASE
    WHEN MONTH(order_date) >= 4 
      THEN CONCAT(YEAR(order_date), '-', YEAR(order_date) + 1)
    ELSE CONCAT(YEAR(order_date) - 1, '-', YEAR(order_date))
  END AS financial_year
FROM orders;
```
