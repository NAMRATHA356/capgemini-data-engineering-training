```python
%sql
SELECT emp_id, name, COALESCE(salary, 0) AS salary
FROM STABLE;
```

```python
%sql
SELECT emp_id,
       NULLIF(salary, 0) AS salary
FROM STABLE;
```

```python
%sql
SELECT order_id,
       NULLIF(discount, 0) AS discount
FROM STABLE;
```

```python
%sql
SELECT order_id,
       amount / NULLIF(discount, 0) AS result
FROM STABLE;
```

```python
%sql
SELECT order_id,
       NULLIF(coupon_code, 'DISC10') AS coupon_code
FROM STABLE;
```

```python
%sql
SELECT emp_id, name,
       COALESCE(salary, 0) + COALESCE(bonus, 0) AS total_earnings
FROM STABLE;
```
