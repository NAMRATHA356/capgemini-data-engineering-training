```python
%sql
SELECT *
FROM Employees e
WHERE EXISTS (
    SELECT 1
    FROM Orders o
    WHERE o.emp_id = e.emp_id
);
```

```python
%sql
SELECT *
FROM Employees
WHERE salary > (
    SELECT MAX(salary)
    FROM Employees
    WHERE department_id = 102
);
```

```python
%sql
WITH cte_highsalary AS (
    SELECT *
    FROM Employees
    WHERE salary > 50000
)
SELECT * FROM cte_highsalary;
```

```python
%sql
WITH cte_avg AS (
    SELECT 
        department_id,
        AVG(salary) AS avg_salary
    FROM Employees
    GROUP BY department_id
)
SELECT * FROM cte_avg;
```

```python
%sql
WITH cte_emp_dept AS (
    SELECT 
        e.emp_id,
        e.name,
        d.dept_name,
        e.salary
    FROM Employees e
    JOIN Departments d
    ON e.department_id = d.dept_id
)
SELECT * FROM cte_emp_dept;
```

```python
%sql
WITH cte_orders AS (
    SELECT 
        emp_id,
        SUM(order_amount) AS total_orders
    FROM Orders
    GROUP BY emp_id
)
SELECT * FROM cte_orders;
```

```python
%sql
WITH cte_avg AS (
    SELECT 
        department_id,
        AVG(salary) AS avg_salary
    FROM Employees
    GROUP BY department_id
)
SELECT e.*
FROM Employees e
JOIN cte_avg a
ON e.department_id = a.department_id
WHERE e.salary > a.avg_salary;
```
